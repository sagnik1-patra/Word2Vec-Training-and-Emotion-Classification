```python
"""Word2Vec and GloVe GoEmotions: full local results, tiny source-only submission.
Run in Jupyter with: %run run_project.py
"""
from pathlib import Path
from collections import Counter
import os, re, json, zipfile, shutil, warnings, random, sys, gc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from gensim.models import Word2Vec
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix, hamming_loss
from sklearn.utils.class_weight import compute_class_weight
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Embedding, GlobalAveragePooling1D, GlobalMaxPooling1D, Dense, Dropout, Conv1D
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam
import joblib

# ===================== USER CONFIGURATION =====================
ROOT = Path(r'C:\Users\sagni\Downloads\Word2Vec Training and Emotion Classification')
OLD_ROOT = Path(r'C:\Users\sagni\Downloads\Training Word2Vec and Emotion Classification using GloVe')
CSV_PATH = ROOT / 'go_emotions_dataset.csv'
if not CSV_PATH.exists():
    CSV_PATH = OLD_ROOT / 'go_emotions_dataset.csv'
# You may instead set CSV_PATH to any existing GoEmotions CSV.
GLOVE_TXT = ROOT / 'assets' / 'glove.6B.100d.txt'
if not GLOVE_TXT.exists() and (OLD_ROOT / 'assets' / 'glove.6B.100d.txt').exists():
    GLOVE_TXT = OLD_ROOT / 'assets' / 'glove.6B.100d.txt'
DOMAIN_CORPUS = ROOT / 'domain_corpus.txt'  # optional: one document per line
RESULTS = ROOT / 'results'
SUBMISSION = ROOT / 'submission'
MAX_ROWS = None              # set to 15000 for a quick test
EPOCHS = 20
BATCH = 64
VOCAB_LIMIT = 12000
SEQ_LEN = 100
DIM = 100
SEED = 42
TRAIN_BILSTM = False         # optional; slower
DOWNLOAD_GLOVE_IF_MISSING = True  # downloads pretrained vectors outside submission
# ==============================================================

random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)
for name in ['graphs', 'csv', 'models', 'reports']:
    (RESULTS / name).mkdir(parents=True, exist_ok=True)
SUBMISSION.mkdir(parents=True, exist_ok=True)
G = RESULTS/'graphs'; C = RESULTS/'csv'; M = RESULTS/'models'; R = RESULTS/'reports'
figures=[]; metrics=[]; histories={}

def saveplot(filename):
    plt.tight_layout(); plt.savefig(G/filename, dpi=110, bbox_inches='tight'); plt.show(); plt.close()
    figures.append(filename)

def savejson(filename, value):
    (R/filename).write_text(json.dumps(value, indent=2, default=str), encoding='utf-8')

def clean(x):
    x = re.sub(r'https?://\S+|www\.\S+', ' ', str(x).lower())
    return ' '.join(re.findall(r"[a-z]+(?:-[a-z]+)*", x))

def eval_model(name, true, pred):
    result={'model':name,'accuracy':float(accuracy_score(true,pred)),
            'macro_f1':float(f1_score(true,pred,average='macro',zero_division=0)),
            'weighted_f1':float(f1_score(true,pred,average='weighted',zero_division=0))}
    metrics.append(result); print(result)

def tsne_2d(vectors):
    n=len(vectors)
    if n<4: return None
    return TSNE(n_components=2,perplexity=min(30,n-1),random_state=SEED).fit_transform(vectors)

if not CSV_PATH.exists():
    raise FileNotFoundError(f'Put go_emotions_dataset.csv at {ROOT} or {OLD_ROOT}')
print('Reading:', CSV_PATH)
df=pd.read_csv(CSV_PATH,nrows=MAX_ROWS)
labels=[c for c in df.columns if c not in ('id','text','example_very_unclear')]
for c in labels:
    df[c]=pd.to_numeric(df[c],errors='coerce').fillna(0).astype(int)
df=df.dropna(subset=['text']).copy()
if 'example_very_unclear' in df:
    df=df[~df.example_very_unclear.astype(str).str.lower().isin(['true','1'])].copy()
df['clean_text']=df.text.map(clean)
df=df[df.clean_text.str.len()>0].copy()
df['label_count']=df[labels].sum(axis=1)
df=df[df.label_count>0].reset_index(drop=True)
print('Rows:',len(df),'Labels:',len(labels))

# EDA
freq=df[labels].sum().sort_values(ascending=False)
freq.rename('count').to_csv(C/'emotion_frequencies.csv')
plt.figure(figsize=(12,9)); sns.barplot(x=freq.values,y=freq.index); plt.title('GoEmotions class distribution'); saveplot('01_class_distribution.png')
lengths=df.clean_text.str.split().map(len)
plt.figure(figsize=(10,5)); sns.histplot(lengths,bins=50); plt.title('Comment lengths'); saveplot('02_comment_length.png')
plt.figure(figsize=(8,5)); sns.countplot(x=df.label_count); plt.title('Number of labels per comment'); saveplot('03_label_counts.png')
pd.DataFrame([{'emotion':label,'text':s} for label in labels for s in df.loc[df[label].eq(1),'text'].head(5)]).to_csv(C/'emotion_examples.csv',index=False)

# Corpus. Do not claim domain-specific requirements are met without checking.
if DOMAIN_CORPUS.exists():
    corpus=[clean(s) for s in DOMAIN_CORPUS.read_text(encoding='utf-8').splitlines()]
    corpus_source='user-provided domain corpus'
else:
    corpus=df.clean_text.tolist(); corpus_source='GoEmotions fallback (not independent domain corpus)'
sentences=[s.split() for s in corpus if s.strip()]
corpus_stats={'source':corpus_source,'documents':len(sentences),'tokens':sum(map(len,sentences))}
corpus_stats['meets_10000_docs']=corpus_stats['documents']>=10000
corpus_stats['meets_1m_tokens']=corpus_stats['tokens']>=1000000
savejson('corpus_stats.json',corpus_stats)
counts=Counter(w for sent in sentences for w in sent)
word_to_index={w:i+1 for i,(w,n) in enumerate(counts.items()) if n>=5}
savejson('word_to_index.json',word_to_index)

# Word2Vec training and hyperparameter experiments
w2v=Word2Vec(sentences=sentences,vector_size=DIM,window=5,min_count=5,workers=4,sg=1,epochs=10,negative=5,seed=SEED)
w2v.save(str(M/'word2vec_custom.model'))
configs=[(50,5,1,5),(100,5,1,10),(200,5,1,10),(300,5,1,5),(100,2,1,10),(100,10,1,10),(100,5,0,10),(100,5,1,20)]
probes=[('happy','joy'),('angry','mad'),('love','like'),('sad','unhappy')]
rows=[]
for dim,win,sg,ep in configs:
    model=Word2Vec(sentences=sentences,vector_size=dim,window=win,min_count=5,workers=4,sg=sg,epochs=ep,seed=SEED)
    scores=[float(model.wv.similarity(a,b)) for a,b in probes if a in model.wv and b in model.wv]
    rows.append({'vector_size':dim,'window':win,'skipgram':sg,'epochs':ep,'probe_similarity':np.mean(scores) if scores else np.nan,'vocab':len(model.wv)})
    del model
exp=pd.DataFrame(rows); exp.to_csv(C/'word2vec_experiments.csv',index=False)
plt.figure(figsize=(10,5)); sns.barplot(data=exp,x='vector_size',y='probe_similarity',hue='skipgram',errorbar=None);plt.title('Word2Vec probe similarity (not accuracy)');saveplot('04_word2vec_hyperparameters.png')
near=[]
for w in ['happy','sad','love','doctor','hospital','king']:
    if w in w2v.wv:
        near.extend([{'query':w,'neighbor':v,'similarity':float(score)} for v,score in w2v.wv.most_similar(w,topn=10)])
pd.DataFrame(near,columns=['query','neighbor','similarity']).to_csv(C/'word2vec_neighbors.csv',index=False)
analogies=[]
for pos,neg in [(['king','woman'],['man']),(['happy','sadness'],['joy']),(['love','hate'],['like'])]:
    if all(w in w2v.wv for w in pos+neg):
        for word,score in w2v.wv.most_similar(positive=pos,negative=neg,topn=5):
            analogies.append({'positive':str(pos),'negative':str(neg),'result':word,'score':float(score)})
pd.DataFrame(analogies,columns=['positive','negative','result','score']).to_csv(C/'word2vec_analogies.csv',index=False)
words=w2v.wv.index_to_key[:500]
coords=tsne_2d(np.array([w2v.wv[w] for w in words]))
if coords is not None:
    plt.figure(figsize=(12,9));plt.scatter(coords[:,0],coords[:,1],s=9)
    for i in range(0,len(words),max(1,len(words)//45)):plt.annotate(words[i],coords[i],fontsize=7)
    plt.title('Word2Vec t-SNE');saveplot('05_word2vec_tsne.png')

# GloVe: keep download outside submission and load only words required for this project.
if not GLOVE_TXT.exists():
    if not DOWNLOAD_GLOVE_IF_MISSING: raise FileNotFoundError('Set GLOVE_TXT to glove.6B.100d.txt')
    import requests
    target=ROOT/'assets';target.mkdir(parents=True,exist_ok=True)
    archive=target/'glove.6B.zip'
    print('Downloading large Stanford GloVe archive ONCE; requires internet and disk space...')
    with requests.get('https://nlp.stanford.edu/data/glove.6B.zip',stream=True,timeout=180) as response:
        response.raise_for_status()
        with archive.open('wb') as f:
            for chunk in response.iter_content(1024*1024):
                if chunk:f.write(chunk)
    with zipfile.ZipFile(archive) as z, z.open('glove.6B.100d.txt') as src, (target/'glove.6B.100d.txt').open('wb') as dst:
        shutil.copyfileobj(src,dst)
    archive.unlink(); GLOVE_TXT=target/'glove.6B.100d.txt'

# Single-label experiment only uses examples with exactly one label.
single=df[df.label_count==1].copy()
single['target']=single[labels].to_numpy().argmax(axis=1)
valid=single.target.value_counts(); valid=valid[valid>=10].index
single=single[single.target.isin(valid)].copy()
X=single.clean_text.to_numpy();y=single.target.to_numpy()
Xtrain,Xtmp,ytrain,ytmp=train_test_split(X,y,test_size=.30,stratify=y,random_state=SEED)
Xval,Xtest,yval,ytest=train_test_split(Xtmp,ytmp,test_size=.5,stratify=ytmp,random_state=SEED)
classes=np.unique(ytrain)
weights=compute_class_weight('balanced',classes=classes,y=ytrain)
class_weights={int(k):float(v) for k,v in zip(classes,weights)}
tok=Tokenizer(num_words=VOCAB_LIMIT,oov_token='<OOV>');tok.fit_on_texts(Xtrain)
def seq(texts,tokenizer=tok):return pad_sequences(tokenizer.texts_to_sequences(texts),maxlen=SEQ_LEN,padding='post',truncating='post')
A,B,T=seq(Xtrain),seq(Xval),seq(Xtest)
n_vocab=min(VOCAB_LIMIT,len(tok.word_index)+1)
needed=set(tok.word_index.keys())|set(w for group in [['joy','happy','excited','thrilled'],['sad','depressed','unhappy','miserable'],['angry','furious','mad','rage'],['love','affection','romance'],['fear','scared','afraid']] for w in group)
# Multi-label tokenizer also needs pretrained vectors, so include all dataset words.
needed.update(w for s in df.clean_text for w in s.split())
glove={}
with GLOVE_TXT.open(encoding='utf-8') as f:
    for line in f:
        parts=line.rstrip().split(' ')
        if len(parts)==DIM+1 and parts[0] in needed:
            glove[parts[0]]=np.asarray(parts[1:],dtype=np.float32)
print('GloVe words loaded:',len(glove))
def matrix(tokenizer,source,size):
    mat=np.zeros((size,DIM),dtype=np.float32);hits=0
    for word,idx in tokenizer.word_index.items():
        if idx<size and word in source:
            mat[idx]=source[word];hits+=1
    return mat,hits
GM,gh=matrix(tok,glove,n_vocab);WM,wh=matrix(tok,w2v.wv,n_vocab)
np.save(M/'glove_matrix.npy',GM);np.save(M/'word2vec_matrix.npy',WM)
savejson('tokenizer.json',tok.to_json())
coverage={'GloVe':100*gh/max(1,n_vocab-1),'Word2Vec':100*wh/max(1,n_vocab-1)}
savejson('embedding_coverage.json',coverage)
plt.figure(figsize=(7,5));sns.barplot(x=list(coverage),y=list(coverage.values()));plt.ylabel('Vocabulary coverage %');saveplot('06_embedding_coverage.png')

def create_net(emb,kind='Dense',trainable=False,output='softmax'):
    net=Sequential([Input(shape=(SEQ_LEN,)),Embedding(len(emb),DIM,weights=[emb],trainable=trainable)])
    if kind=='CNN':net.add(Conv1D(128,5,activation='relu'));net.add(GlobalMaxPooling1D());net.add(Dense(128,activation='relu'))
    elif kind=='BiLSTM':
        from tensorflow.keras.layers import Bidirectional,LSTM
        net.add(Bidirectional(LSTM(64)));net.add(Dense(64,activation='relu'))
    else:net.add(GlobalAveragePooling1D());net.add(Dense(128,activation='relu'));net.add(Dense(64,activation='relu'))
    net.add(Dropout(.5));net.add(Dense(len(labels),activation=output))
    net.compile(optimizer=Adam(.001),loss='binary_crossentropy' if output=='sigmoid' else 'sparse_categorical_crossentropy',metrics=['binary_accuracy' if output=='sigmoid' else 'accuracy'])
    return net

def train_single(name,emb,kind='Dense',trainable=False):
    tf.keras.backend.clear_session();net=create_net(emb,kind,trainable)
    h=net.fit(A,ytrain,validation_data=(B,yval),epochs=EPOCHS,batch_size=BATCH,class_weight=class_weights,callbacks=[EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)],verbose=1)
    histories[name]={k:list(map(float,v)) for k,v in h.history.items()}
    plt.figure(figsize=(8,4));plt.plot(h.history['loss'],label='train');plt.plot(h.history['val_loss'],label='validation');plt.legend();plt.title(name+' loss');saveplot(name+'_loss.png')
    plt.figure(figsize=(8,4));plt.plot(h.history['accuracy'],label='train');plt.plot(h.history['val_accuracy'],label='validation');plt.legend();plt.title(name+' accuracy');saveplot(name+'_accuracy.png')
    prob=net.predict(T,verbose=0);pred=prob.argmax(axis=1)
    eval_model(name,ytest,pred)
    pd.DataFrame({'text':Xtest,'actual':[labels[i] for i in ytest],'predicted':[labels[i] for i in pred],'confidence':prob.max(axis=1)}).to_csv(C/f'{name}_predictions.csv',index=False)
    net.save(str(M/f'{name}.h5'));return pred

predictions={}
predictions['GloVe_Dense']=train_single('GloVe_Dense',GM)
predictions['GloVe_CNN']=train_single('GloVe_CNN',GM,'CNN')
predictions['Word2Vec_Dense']=train_single('Word2Vec_Dense',WM)
predictions['GloVe_FineTuned']=train_single('GloVe_FineTuned',GM,trainable=True)
if TRAIN_BILSTM:predictions['GloVe_BiLSTM']=train_single('GloVe_BiLSTM',GM,'BiLSTM')

# TF-IDF baselines
vectorizer=TfidfVectorizer(max_features=15000,ngram_range=(1,2),min_df=2,sublinear_tf=True)
Xt=vectorizer.fit_transform(Xtrain);Xe=vectorizer.transform(Xtest)
joblib.dump(vectorizer,M/'tfidf.pkl',compress=3)
for name,model in [('TFIDF_LogisticRegression',LogisticRegression(max_iter=500,class_weight='balanced')),
                   ('TFIDF_RandomForest',RandomForestClassifier(n_estimators=100,max_depth=25,n_jobs=-1,random_state=SEED))]:
    model.fit(Xt,ytrain);pred=model.predict(Xe);predictions[name]=pred;eval_model(name,ytest,pred)
    joblib.dump(model,M/f'{name}.pkl',compress=3)
comparison=pd.DataFrame(metrics);comparison.to_csv(C/'model_comparison.csv',index=False)
for col,filename in [('accuracy','07_model_accuracy.png'),('macro_f1','08_model_macro_f1.png')]:
    plt.figure(figsize=(11,5));sns.barplot(data=comparison,x=col,y='model');plt.title('Model comparison: '+col);saveplot(filename)
# Select using validation in a rigorous study; here this is a descriptive best-on-test summary.
best=comparison.sort_values('macro_f1',ascending=False).iloc[0]['model'];best_pred=predictions[best]
report=classification_report(ytest,best_pred,labels=list(range(len(labels))),target_names=labels,output_dict=True,zero_division=0)
pd.DataFrame(report).T.to_csv(C/'classification_report.csv')
cm=confusion_matrix(ytest,best_pred,labels=list(range(len(labels))))
plt.figure(figsize=(16,13));sns.heatmap(cm,xticklabels=labels,yticklabels=labels,cmap='Blues');plt.title('Confusion matrix: '+best);saveplot('09_confusion_matrix.png')
f1s=pd.DataFrame(report).T.loc[labels,'f1-score'].sort_values()
plt.figure(figsize=(10,9));f1s.plot.barh();plt.title('Per-class F1');saveplot('10_per_class_f1.png')
errors=pd.DataFrame({'text':Xtest,'actual':[labels[i] for i in ytest],'predicted':[labels[i] for i in best_pred]})
errors=errors[errors.actual!=errors.predicted];errors.to_csv(C/'misclassified_examples.csv',index=False)
pairs=errors.groupby(['actual','predicted']).size().reset_index(name='count').sort_values('count',ascending=False)
pairs.to_csv(C/'confused_pairs.csv',index=False)
top=pairs.head(15).copy();top['pair']=top.actual+' → '+top.predicted
plt.figure(figsize=(11,6));sns.barplot(data=top,x='count',y='pair');plt.title('Most confused pairs');saveplot('11_confused_pairs.png')
analysis=pd.DataFrame({'length':[len(t.split()) for t in Xtest],'correct':(ytest==best_pred).astype(int)})
analysis['bucket']=pd.cut(analysis.length,[0,5,10,20,40,100000],include_lowest=True)
plt.figure(figsize=(10,5));analysis.groupby('bucket',observed=True).correct.mean().plot.bar();plt.title('Accuracy by length');saveplot('12_length_accuracy.png')

# GloVe emotion word space
emotion_words={'happy':['joy','happy','excited','thrilled'],'sad':['sad','depressed','unhappy','miserable'],'anger':['angry','furious','mad','rage'],'love':['love','affection','romance'],'fear':['fear','scared','afraid']}
selected=[(group,word) for group,words in emotion_words.items() for word in words if word in glove]
if len(selected)>=4:
    vectors=np.array([glove[w] for _,w in selected]); xy=tsne_2d(vectors)
    plt.figure(figsize=(10,7))
    for group in emotion_words:
        ids=[i for i,(g,_) in enumerate(selected) if g==group]
        if ids:plt.scatter(xy[ids,0],xy[ids,1],label=group)
    for i,(_,word) in enumerate(selected):plt.annotate(word,xy[i])
    plt.legend();plt.title('GloVe emotion t-SNE');saveplot('13_glove_tsne.png')
    xyz=PCA(n_components=3).fit_transform(vectors)
    fig=plt.figure(figsize=(10,7));ax=fig.add_subplot(111,projection='3d')
    for group in emotion_words:
        ids=[i for i,(g,_) in enumerate(selected) if g==group]
        if ids:ax.scatter(xyz[ids,0],xyz[ids,1],xyz[ids,2],label=group)
    ax.legend();ax.set_title('GloVe emotion PCA 3D');saveplot('14_glove_pca_3d.png')

# Class centroids based on average document GloVe vectors
centroids=[]
for label in labels:
    vectors=[]
    for text in df.loc[df[label]==1,'clean_text'].head(3000):
        vs=[glove[w] for w in text.split() if w in glove]
        if vs:vectors.append(np.mean(vs,axis=0))
    centroids.append(np.mean(vectors,axis=0) if vectors else np.zeros(DIM))
centroids=np.array(centroids);xy=PCA(n_components=2).fit_transform(centroids)
plt.figure(figsize=(11,8));plt.scatter(xy[:,0],xy[:,1]);
for i,label in enumerate(labels):plt.annotate(label,xy[i],fontsize=8)
plt.title('Emotion centroids');saveplot('15_emotion_centroids.png')
sim=cosine_similarity(centroids);pd.DataFrame(sim,index=labels,columns=labels).to_csv(C/'emotion_similarity.csv')
plt.figure(figsize=(13,11));sns.heatmap(sim,xticklabels=labels,yticklabels=labels,cmap='coolwarm',vmin=-1,vmax=1);plt.title('Emotion similarity');saveplot('16_emotion_similarity.png')

# Multi-label model: separate training-only tokenizer avoids leakage.
Xm=df.clean_text.to_numpy();Ym=df[labels].to_numpy(dtype=np.float32)
Xmtrain,Xmtmp,Ymtrain,Ymtmp=train_test_split(Xm,Ym,test_size=.30,random_state=SEED)
Xmval,Xmtest,Ymval,Ymtest=train_test_split(Xmtmp,Ymtmp,test_size=.5,random_state=SEED)
mtok=Tokenizer(num_words=VOCAB_LIMIT,oov_token='<OOV>');mtok.fit_on_texts(Xmtrain)
mvocab=min(VOCAB_LIMIT,len(mtok.word_index)+1)
MM,_=matrix(mtok,glove,mvocab)
def mseq(t):return pad_sequences(mtok.texts_to_sequences(t),maxlen=SEQ_LEN,padding='post',truncating='post')
Am,Bm,Tm=mseq(Xmtrain),mseq(Xmval),mseq(Xmtest)
tf.keras.backend.clear_session();multi=create_net(MM,'CNN',False,'sigmoid')
h=multi.fit(Am,Ymtrain,validation_data=(Bm,Ymval),epochs=EPOCHS,batch_size=BATCH,callbacks=[EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)],verbose=1)
plt.figure(figsize=(8,4));plt.plot(h.history['loss'],label='train');plt.plot(h.history['val_loss'],label='validation');plt.legend();plt.title('Multi-label loss');saveplot('17_multilabel_loss.png')
vp=multi.predict(Bm,verbose=0);thresholds=np.arange(.15,.61,.05)
scores=[f1_score(Ymval,(vp>=t).astype(int),average='micro',zero_division=0) for t in thresholds]
threshold=float(thresholds[np.argmax(scores)])
probs=multi.predict(Tm,verbose=0);mp=(probs>=threshold).astype(int)
multimetrics={'threshold':threshold,'hamming_loss':float(hamming_loss(Ymtest,mp))}
for average in ['micro','macro','weighted']:
    multimetrics[f'{average}_f1']=float(f1_score(Ymtest,mp,average=average,zero_division=0))
for k in [1,3,5]:
    topidx=np.argsort(-probs,axis=1)[:,:k]
    hits=np.take_along_axis(Ymtest,topidx,axis=1).sum(axis=1)
    multimetrics[f'precision_at_{k}']=float(np.mean(hits/k))
    multimetrics[f'recall_at_{k}']=float(np.mean(hits/np.maximum(1,Ymtest.sum(axis=1))))
savejson('multilabel_metrics.json',multimetrics);savejson('multilabel_tokenizer.json',mtok.to_json())
multi.save(str(M/'GloVe_MultiLabel.h5'))
pd.DataFrame({'text':Xmtest,'actual':[','.join(np.array(labels)[row.astype(bool)]) for row in Ymtest],
              'predicted':[','.join(np.array(labels)[row.astype(bool)]) for row in mp]}).to_csv(C/'multilabel_predictions.csv',index=False)

plt.figure(figsize=(10,5));pd.DataFrame({'frequency':freq.reindex(labels).values,'f1':[report[k]['f1-score'] for k in labels]}).plot.scatter(x='frequency',y='f1');plt.title('Class frequency vs F1');saveplot('18_frequency_vs_f1.png')
lex=pd.DataFrame({'text':Xtest,'correct':(ytest==best_pred).astype(int)})
lex['richness']=lex.text.map(lambda t:len(set(t.split()))/max(1,len(t.split())))
lex['bucket']=pd.cut(lex.richness,np.linspace(0,1,6),include_lowest=True)
plt.figure(figsize=(10,5));lex.groupby('bucket',observed=True).correct.mean().plot.bar();plt.title('Lexical richness vs accuracy');saveplot('19_lexical_richness.png')

savejson('training_histories.json',histories)
summary={'rows':len(df),'labels':labels,'corpus':corpus_stats,'embedding_coverage':coverage,
         'model_comparison':metrics,'best_by_test_macro_f1':best,'multi_label':multimetrics,'graphs':figures}
savejson('summary.json',summary)
report_text=f'''# Technical Report: Word2Vec and GloVe Emotion Classification\n\n## Dataset and preprocessing\n{len(df)} GoEmotions comments retained; {len(labels)} labels. Lowercasing, URL removal, tokenization and filtering were used.\n\n## Word2Vec training\n{corpus_stats}\nSkip-gram and CBOW hyperparameters were compared. Probe similarity is NOT downstream accuracy.\n\n## GloVe and architectures\nGloVe 6B 100d; coverage: {coverage}. Dense, CNN, Word2Vec Dense, and fine-tuned Dense models were trained.\n\n## Results\n{comparison.to_string(index=False)}\n\n## Multi-label evaluation\n{multimetrics}\n\n## Error analysis\n{pairs.head(10).to_string(index=False)}\n\n## Limitations\nAn independent 1-million-token domain corpus is required for full corpus compliance. WordSim-353/SimLex-999 and GloVe 300d are not included. The best-on-test summary is descriptive rather than a valid model-selection protocol. Results vary with random seed and hardware.\n'''
(R/'technical_report.md').write_text(report_text,encoding='utf-8')

# ------------------- SLIM SUBMISSION -------------------
# Never put the whole results directory, original CSV, pretrained GloVe, or .h5/.pkl/.npy into the submission.
# Copy only source artifacts and ONE explicitly requested visualization.
source=Path(__file__).resolve() if '__file__' in globals() else ROOT/'run_project.py'
if source.exists():shutil.copy2(source,SUBMISSION/'run_project.py')
readme='''# Word2Vec Training and Emotion Classification using GloVe\n\n## Overview\nGoEmotions NLP project comparing custom Word2Vec embeddings, pretrained GloVe, Dense/CNN classifiers, and TF-IDF baselines. Includes multi-label classification and embedding/error analysis.\n\n![Emotion class distribution](graphs/01_class_distribution.png)\n\n## Setup\nInstall dependencies using `pip install -r requirements.txt`. Place `go_emotions_dataset.csv` in the configured project directory (not inside the submission ZIP). Run `run_project.py` or `notebook.ipynb`. The script downloads Stanford GloVe if no local copy exists; this large asset is intentionally excluded from the ZIP.\n\n## Results\nAll figures, CSVs, models, and reports are saved locally in `results/`, not submitted.\n\n## Important limitations\nThe GoEmotions fallback corpus is not an independent domain-specific corpus. External WordSim-353/SimLex-999 benchmarking and 300d GloVe experiments are not implemented.\n'''
(SUBMISSION/'README.md').write_text(readme,encoding='utf-8')
(SUBMISSION/'requirements.txt').write_text('numpy\npandas\nscipy\nscikit-learn\ngensim\ntensorflow\nmatplotlib\nseaborn\nrequests\njoblib\nnbformat\n',encoding='utf-8')
(SUBMISSION/'graphs').mkdir(exist_ok=True)
shutil.copy2(G/'01_class_distribution.png',SUBMISSION/'graphs'/'01_class_distribution.png')
try:
    import nbformat
    nb=nbformat.v4.new_notebook(cells=[
        nbformat.v4.new_markdown_cell('# Word2Vec and GloVe Emotion Classification\nClean reproducible notebook. See README.md.'),
        nbformat.v4.new_code_cell('%run run_project.py')
    ])
    nbformat.write(nb,SUBMISSION/'notebook.ipynb')
except ImportError:
    print('Install nbformat to generate notebook.ipynb')
zip_path=ROOT/'Word2Vec_GloVe_Submission.zip'
allow={'run_project.py','README.md','requirements.txt','notebook.ipynb','graphs/01_class_distribution.png'}
with zipfile.ZipFile(zip_path,'w',zipfile.ZIP_DEFLATED,compresslevel=9) as z:
    for file in SUBMISSION.rglob('*'):
        if file.is_file() and file.relative_to(SUBMISSION).as_posix() in allow:
            z.write(file,file.relative_to(SUBMISSION).as_posix())
print('\nALL RESULTS:',RESULTS)
print('SLIM SUBMISSION:',zip_path,'Size MB:',round(zip_path.stat().st_size/1024**2,3))
print('ZIP FILES:',zipfile.ZipFile(zip_path).namelist())
```
